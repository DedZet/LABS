## 1. Вектор

### 1.1 Создайте вектор S1 со значениями от 1 до 10 всеми возможными способами

```python
S1_list = list(range(1, 11))
S1_tuple = tuple(range(1, 11))
S1_np = np.arange(1, 11)
S1_pd = pd.Series(range(1, 11))

print(S1_list)
print(S1_tuple)
print(S1_np)
print(S1_pd)
```

### 1.2 Создайте вектор S2 со значениями от 1 до 10 с шагом приращения 0,5

```python
S2 = np.arange(1, 10.5, 0.5)
print(S2)
```

### 1.3 Объедините векторы S1 и S2 в вектор S

```python
S = np.concatenate([S1_np, S2])
print(S)
```

### 1.4 Выведите на экран 3, 4 и 5 значения вектора S

```python
print(S[2:5])
```

### 1.5 Выберите только первое и четвертое значения вектора S

```python
print(S[[0, 3]])
```

## 2. Таблица

### 2.1 Создайте текстовые векторы с названиями городов (City) и пола (Sex), а также вектор со значениями численности представителей каждого пола (Number)

```python
cities = ["Washington", "New-York", "London", "Paris", "Tokyo", "Osaka"]
numbers = [[120, 130], [240, 250], [160, 450], [204, 270], [587, 652], [201, 321]]
City = [c for c in cities for _ in range(2)]
Sex = ["male", "female"] * 6
Number = [n for pair in numbers for n in pair]
print(City)
print(Sex)
print(Number)
```

### 2.3 Объедините эти три вектора в одну таблицу данных CITY с заголовками столбцов (City, Sex, Number)

```python
CITY = pd.DataFrame({"City": City, "Sex": Sex, "Number": Number})
CITY
```

### 2.4 Визуализируйте содержимое таблицы CITY

```python
display(CITY)
```

### 2.5 Выведите имена переменных, входящих в таблицу данных CITY

```python
print(CITY.columns.tolist())
```

### 2.6 Извлеките элементы 1-3 из столбца Number таблицы CITY

```python
print(CITY["Number"].iloc[0:3])
```

### 2.7 Извлеките все значения численности населения таблицы CITY, превышающие 10000

```python
print(CITY[CITY["Number"] > 10000])
```

### 2.8 Извлеките все значения численности мужского населения таблицы CITY

```python
print(CITY.loc[CITY["Sex"] == "male", "Number"])
```

### 2.9 Для просмотра трех первых и двух последних значений каждой переменной, входящей в состав таблицы данных CITY используйте функции head() и tail() соответственно

```python
print(CITY.head(3))
print(CITY.tail(2))
```

## 6. Фактор

### 6.1 Создайте вектор из 6-ти объектов, каждый из которых принадлежит одному из трех классов: Yes, No, Parhaps. Преобразуйте его в фактор

```python
f1 = pd.Categorical(["Yes", "No", "Parhaps", "Yes", "No", "Parhaps"], categories=["Yes", "No", "Parhaps"])
print(f1)
print(f1.categories.tolist())
print(f1.codes)
```
